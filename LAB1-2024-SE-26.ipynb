```python
def caught_speeding(speed,is_birthday):
    if is_birthday:
        speed=speed-5
        if speed<=60:
            return "no ticket"
        elif speed<=80:
            return "small ticket"
        else:
            return "big ticket"

print(caught_speeding(83,True))
print(caught_speeding(60,False))
```
