## Task 3 · Randomness and reproducibility
- **(a)** Set `np.random.seed(2026)`, then generate one random number between 0 and 1.
- **(b)** Generate 25 numbers from the standard normal distribution.
- **(c)** Set `np.random.seed(42)` and roll 10,000 Ludo dice with `np.random.randint`. Count each face with `np.unique(..., return_counts=True)`.
- **(d)** Estimate P(six) and compare it with 1/6.
- **(e)** Count how many times two sixes came in a row (no loops).

```python
# T3: Randomness and reproducibility
np.random.seed(2026)
one_random_number = np.random.random()
print("T3a: One random number:", one_random_number)

normal_numbers = np.random.randn(25)
print("T3b: 25 standard normal numbers:", normal_numbers)

np.random.seed(42)
dice = np.random.randint(1, 7, size=10000)
faces, face_counts = np.unique(dice, return_counts=True)
print("T3c: Dice faces:", faces)
print("T3c: Counts for each face:", face_counts)

six_count = np.sum(dice == 6)
six_probability = six_count / dice.size
print("T3d: Estimated probability of a six:", six_probability)
print("T3d: Theoretical probability of a six:", 1 / 6)

two_sixes_in_row = np.sum((dice[:-1] == 6) & (dice[1:] == 6))
print("T3e: Number of adjacent pairs of sixes:", two_sixes_in_row)
```

## Task 6 · Census 2023 (real data, PBS)
Using `units`, `pop_2023` and `pop_2017`:

- **(c)** The most populous unit.

**Interpret:** in 1–2 lines, what do (e) and (f) say? Compare AJK with the rest.

## Task 7 · Climate of five cities (real data, PMD 1991–2020 normals)
Using `tmax` (mean daily maximum °C) and `rain` (mm), rows = `cities`, columns = `months`:

- **(h)** The wettest city in the monsoon, and what % of each city's yearly rain falls in the monsoon.

**Interpret:** in 1–2 lines, which result surprised you, and why?

```python
# T7: Climate of five cities
print("T7a: Shape of tmax:", tmax.shape)

annual_mean = np.mean(tmax, axis=1)
print("T7b: Cities:", cities)
print("T7b: Annual mean maximum temperature (degrees C):", np.round(annual_mean, 2))

hottest_indices = np.argmax(tmax, axis=1)
print("T7c: Hottest month for each city:", months[hottest_indices])
print("T7c: Hottest temperature for each city (degrees C):", tmax[np.arange(len(cities)), hottest_indices])

monthly_city_average = np.mean(tmax, axis=0)
print("T7d: Five-city average for each month (degrees C):", np.round(monthly_city_average, 2))

months_above_35 = np.sum(tmax > 35, axis=1)
print("T7e: Number of months above 35 degrees C for each city:", months_above_35)

lahore_fahrenheit = tmax[0] * 9 / 5 + 32
print("T7f: Lahore temperatures in degrees F:", np.round(lahore_fahrenheit, 2))

monsoon_rain = rain[:, 6:9]
monsoon_total = np.sum(monsoon_rain, axis=1)
yearly_rain = np.sum(rain, axis=1)
monsoon_share = monsoon_total / yearly_rain * 100
wettest_monsoon_index = np.argmax(monsoon_total)
print("T7g: Monsoon rainfall for Jul, Aug, Sep (mm):\n", monsoon_rain)
print("T7g: Total monsoon rainfall for each city (mm):", np.round(monsoon_total, 1))
print("T7h: Wettest monsoon city:", cities[wettest_monsoon_index], "-", round(monsoon_total[wettest_monsoon_index], 1), "mm")
print("T7h: Percentage of yearly rain falling in monsoon (%):", np.round(monsoon_share, 2))
```

## Task 8 · Inflation FY2024 vs FY2025 (real data, Pakistan Economic Survey 2024-25)

- **(a)** Average inflation of each year.
- **(b)** FY2024 months with inflation above 25 %.

- **(e)** Label each FY2025 month "Low" (< 5), "Moderate" (5–10) or "High" (≥ 10) with `np.where`.

**Interpret:** in 1–2 lines, what happened to prices in Pakistan between the two years?

```python
# T8: Inflation FY2024 vs FY2025
average_inflation = np.mean(cpi, axis=1)
above_25_months = cpi_months[cpi[0] > 25]
monthly_drop = cpi[0] - cpi[1]
biggest_drop_index = np.argmax(monthly_drop)
fy2025_labels = np.where(cpi[1] < 5, "Low", np.where(cpi[1] < 10, "Moderate", "High"))

print("T8a: Average inflation in FY2024 (%):", round(average_inflation[0], 2))
print("T8a: Average inflation in FY2025 (%):", round(average_inflation[1], 2))
print("T8b: FY2024 months above 25% inflation:", above_25_months)
print("T8c: Months:", cpi_months)
print("T8c: Monthly drop in inflation (percentage points):", np.round(monthly_drop, 1))
print("T8d: Month with the biggest drop:", cpi_months[biggest_drop_index], "-", round(monthly_drop[biggest_drop_index], 1), "percentage points")
print("T8e: FY2025 labels for Jul-Apr:", fy2025_labels)
```

## Task 9 · Grace marks
Using `marks`:
- **(a)** Add 5 grace marks but cap at 100 (`np.clip`).
- **(b)** Pass % (≥ 50) after grace.
- **(c)** Index of students who passed **only because of** the grace marks.
- **(d)** z-score of each student after grace.

```python
# T9: Grace marks
marks_after_grace = np.clip(marks + 5, 0, 100)
pass_percentage = np.sum(marks_after_grace >= 50) / marks_after_grace.size * 100
passed_only_with_grace = np.where((marks < 50) & (marks_after_grace >= 50))[0]
z_scores = (marks_after_grace - np.mean(marks_after_grace)) / np.std(marks_after_grace)

print("T9a: Marks after adding 5 grace marks (maximum 100):", marks_after_grace)
print("T9b: Percentage passing after grace:", round(pass_percentage, 2), "%")
print("T9c: Student indexes who passed only because of grace marks:", passed_only_with_grace)
print("T9d: Z-scores after grace:", np.round(z_scores, 3))
```

- **(f)** Print its shape, dtype and number of white pixels.

*Optional:* `import matplotlib.pyplot as plt; plt.imshow(img, cmap='gray')`
