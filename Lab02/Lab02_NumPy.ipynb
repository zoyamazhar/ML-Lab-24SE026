## Task 3 · Randomness and reproducibility
- **(a)** Set `np.random.seed(2026)`, then generate one random number between 0 and 1.
- **(b)** Generate 25 numbers from the standard normal distribution.
- **(c)** Set `np.random.seed(42)` and roll 10,000 Ludo dice with `np.random.randint`. Count each face with `np.unique(..., return_counts=True)`.
- **(d)** Estimate P(six) and compare it with 1/6.
- **(e)** Count how many times two sixes came in a row (no loops).

```python
# T3a: Set the seed and generate one random number from 0 to 1
np.random.seed(2026)
random_number = np.random.rand()
print("T3a: Random number:", random_number)

# T3b: Generate 25 numbers from the standard normal distribution
normal_numbers = np.random.randn(25)
print("T3b: Standard normal numbers:", normal_numbers)

# T3c: Set the seed and simulate 10,000 dice rolls
np.random.seed(42)
dice = np.random.randint(1, 7, 10000)
faces, counts = np.unique(dice, return_counts=True)
print("T3c: Dice faces:", faces)
print("T3c: Counts for each face:", counts)

# T3d: Estimate the probability of rolling a six and compare with 1/6
six_count = np.sum(dice == 6)
six_probability = six_count / len(dice)
print("T3d: Estimated probability of six:", six_probability)
print("T3d: Theoretical probability of six (1/6):", 1 / 6)
print("T3d: Absolute difference:", abs(six_probability - 1 / 6))

# T3e: Count pairs of sixes next to each other, without using a loop
double_sixes = np.sum((dice[:-1] == 6) & (dice[1:] == 6))
print("T3e: Consecutive pairs of sixes:", double_sixes)
```

## Task 8 · Inflation FY2024 vs FY2025 (real data, Pakistan Economic Survey 2024-25)

- **(e)** Label each FY2025 month "Low" (< 5), "Moderate" (5–10) or "High" (≥ 10) with `np.where`.

**Interpret:** in 1–2 lines, what happened to prices in Pakistan between the two years?

```python
# T8a: Average inflation for FY2024 and FY2025
average_inflation = np.mean(cpi, axis=1)
print("T8a: Average inflation in FY2024 (%):", round(average_inflation[0], 2))
print("T8a: Average inflation in FY2025 (%):", round(average_inflation[1], 2))

# T8b: Find FY2024 months with inflation above 25 percent
high_inflation_months = cpi_months[cpi[0] > 25]
print("T8b: FY2024 months above 25% inflation:", high_inflation_months)

# T8c: Calculate the month-by-month drop in inflation
inflation_drop = cpi[0] - cpi[1]
print("T8c: Drop in inflation (percentage points):", dict(zip(cpi_months.tolist(), np.round(inflation_drop, 1).tolist())))

# T8d: Find the month with the biggest drop
biggest_drop_index = np.argmax(inflation_drop)
print("T8d: Month with the biggest drop:", cpi_months[biggest_drop_index])
print("T8d: Biggest drop (percentage points):", round(inflation_drop[biggest_drop_index], 1))

# T8e: Label each FY2025 month Low, Moderate or High using np.where
inflation_labels = np.where(
    cpi[1] < 5,
    "Low",
    np.where(cpi[1] < 10, "Moderate", "High")
)
print("T8e: FY2025 inflation labels:", dict(zip(cpi_months.tolist(), inflation_labels.tolist())))
```

## Task 9 · Grace marks
Using `marks`:
- **(a)** Add 5 grace marks but cap at 100 (`np.clip`).
- **(b)** Pass % (≥ 50) after grace.
- **(c)** Index of students who passed **only because of** the grace marks.
- **(d)** z-score of each student after grace.

```python
# T9a: Add 5 grace marks and cap the marks at 100
marks_after_grace = np.clip(marks + 5, 0, 100)
print("T9a: Marks after grace marks:", marks_after_grace)

# T9b: Calculate the percentage of students who passed (50 or above)
passed_count = np.sum(marks_after_grace >= 50)
pass_percentage = (passed_count / len(marks_after_grace)) * 100
print("T9b: Passing percentage after grace (%):", round(pass_percentage, 2))

# T9c: Find indexes of students who passed only because of grace marks
grace_passed_students = np.where((marks < 50) & (marks_after_grace >= 50))[0]
print("T9c: Indexes of students who passed only because of grace:", grace_passed_students)

# T9d: Calculate each student's z-score after grace marks
z_scores = (marks_after_grace - np.mean(marks_after_grace)) / np.std(marks_after_grace)
print("T9d: Z-scores after grace marks:", np.round(z_scores, 3))
```

*Optional:* `import matplotlib.pyplot as plt; plt.imshow(img, cmap='gray')`

```python
# T10a: Create an 8 x 8 image and draw the number 7 from the lecture
img = np.zeros((8, 8), dtype=int)
img[1, 1:7] = 255       # White horizontal line
img[2:7, 5] = 255       # White vertical line
img[4, 3:7] = 128       # Grey line across row 4
print("T10a: Original image:\n", img)

# T10b: Flip the image from left to right
flipped_img = np.fliplr(img)
print("T10b: Flipped image:\n", flipped_img)

# T10c: Invert the image
inverted_img = 255 - img
print("T10c: Inverted image:\n", inverted_img)

# T10d: Crop rows 1-4 and columns 3-6 inclusive (Python stop is exclusive: rows 1:5, columns 3:7)
cropped_img = img[1:5, 3:7]
print("T10d: Cropped image (rows 1-4, columns 3-6), shape =", cropped_img.shape, "\n", cropped_img)

# T10e: Brighten the image by 80 and cap values at 255
bright_img = np.clip(img + 80, 0, 255)
print("T10e: Brightened image:\n", bright_img)

# T10f: Print the original image's shape, data type and number of white pixels
print("T10f: Image shape:", img.shape)
print("T10f: Image data type:", img.dtype)
print("T10f: Number of white pixels:", np.sum(img == 255))
```
