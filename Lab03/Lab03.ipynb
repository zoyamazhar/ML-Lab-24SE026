## Task 3 · Clean the data: the six types of problem
Fix the problems **in this order** (text first, then duplicates):
- **(a) Missing values:** count them per column. Drop ads with no `Brand` or no `Model`; fill missing `Registered City`, `Fuel` and `Transaction Type` with `"Unknown"`; leave `Year` and `KMs Driven` as NaN. Print how many rows were removed.
- **(b) Formatting:** remove extra spaces and fix the case of city names (`.str.strip().str.title()`); remove `"Rs"` and commas from `Price`.
- **(c) Data types:** convert `Price` to `int64` and `Year` to `Int64`.
- **(d) Inconsistent categories:** `"Sukkar"` → `"Sukkur"` and `"Corrolla Altis"` → `"Corolla Altis"`. Print the number of different cities before and after.
- **(e) Duplicates:** count exact duplicate ads, drop them, print the new row count.
- **(f) Invalid values:** set `KMs Driven` to NaN when it is below 100, above 1,000,000, or the placeholder 12345 / 123456. Drop ads with the fake prices 12,345,678 or 87,654,321, and Suzuki/Daihatsu ads above 5,000,000.
- **(g) Self-check:** your final shape must be **(20484, 9)**.

```python
# T3: Clean the data in the required order

# T3a: Count missing values and remove rows without Brand or Model
print("T3a: Missing values before cleaning:")
print(df.isna().sum())

rows_before = len(df)
df = df.dropna(subset=["Brand", "Model"])
rows_removed = rows_before - len(df)
print("T3a: Rows removed because Brand or Model was missing:", rows_removed)

df["Registered City"] = df["Registered City"].fillna("Unknown")
df["Fuel"] = df["Fuel"].fillna("Unknown")
df["Transaction Type"] = df["Transaction Type"].fillna("Unknown")

# T3b: Fix text formatting and clean Price
df["Registered City"] = df["Registered City"].astype("string").str.strip().str.title()
df["Price"] = df["Price"].astype(str).str.replace("Rs", "", regex=False)
df["Price"] = df["Price"].str.replace(",", "", regex=False).str.strip()

# T3c: Convert columns to the required data types
df["Price"] = pd.to_numeric(df["Price"], errors="coerce").astype("int64")
df["Year"] = df["Year"].astype("Int64")

# T3d: Fix the two inconsistent categories
cities_before = df["Registered City"].nunique()
df["Registered City"] = df["Registered City"].replace({"Sukkar": "Sukkur"})
df["Model"] = df["Model"].replace({"Corrolla Altis": "Corolla Altis"})
cities_after = df["Registered City"].nunique()
print("T3d: Different city names before correction:", cities_before)
print("T3d: Different city names after correction:", cities_after)

# T3e: Count and remove exact duplicate ads
duplicate_count = df.duplicated().sum()
print("T3e: Exact duplicate ads:", duplicate_count)
df = df.drop_duplicates()
print("T3e: Rows after removing duplicates:", len(df))

# T3f: Replace invalid kilometre values with missing values
bad_km = (df["KMs Driven"] < 100) | (df["KMs Driven"] > 1000000) | (df["KMs Driven"].isin([12345, 123456]))
df.loc[bad_km, "KMs Driven"] = np.nan

# Remove fake prices and Suzuki/Daihatsu ads above the given limit
df = df[~df["Price"].isin([12345678, 87654321])]
df = df[~(df["Brand"].isin(["Suzuki", "Daihatsu"]) & (df["Price"] > 5000000))]

print("T3f: Cleaning invalid values completed.")
print("T3g: Final shape check:", df.shape)
print("T3g: Expected shape is (20484, 9).")
```

## Task 4 · Analyse
- **(a)** New column `Price_lakh` = price in lakh rupees (÷ 100,000).
- **(b)** New column `Age` = 2020 − `Year`.
- **(c)** The 5 most common brands.
- **(d)** The 5 cities with the most ads.
- **(e)** Median `Price_lakh` for each fuel type, highest first.
- **(f)** Median `Price_lakh` of a **Corolla GLI** for each year 2010–2018. How much value does it lose per year?
- **(g)** The 5 most expensive ads. Does one of them look wrong? Why?
- **(h)** Number of ads and median price for Cash vs Installment/Leasing. Interpret in 1–2 lines.

```python
# T4: Analyse the cleaned car data

# T4a: Convert price into lakh rupees
df["Price_lakh"] = df["Price"] / 100000
print("T4a: First 5 values of Price_lakh:")
print(df["Price_lakh"].head())

# T4b: Calculate car age using the year 2020 as required
df["Age"] = 2020 - df["Year"]
print("T4b: First 5 values of Age:")
print(df["Age"].head())

# T4c: Five most common brands
print("T4c: Five most common brands:")
print(df["Brand"].value_counts().head(5))

# T4d: Five cities with the most advertisements
print("T4d: Advertisement counts including Unknown:")
print(df["Registered City"].value_counts().head(6))

print("T4d: Five actual cities with the most ads (excluding Unknown):")
city_counts = df.loc[df["Registered City"] != "Unknown", "Registered City"].value_counts()
print(city_counts.head(5))

# T4e: Median price in lakh for each fuel type, highest first
print("T4e: Median Price_lakh by fuel type, highest first:")
print(df.groupby("Fuel")["Price_lakh"].median().sort_values(ascending=False))

# T4f: Median price of Corolla GLI for each year from 2010 to 2018
corolla = df[(df["Model"] == "Corolla GLI") & (df["Year"] >= 2010) & (df["Year"] <= 2018)]
corolla_medians = corolla.groupby("Year")["Price_lakh"].median().sort_index()
print("T4f: Median price in lakh for Corolla GLI, 2010-2018:")
print(corolla_medians)
if 2010 in corolla_medians.index and 2018 in corolla_medians.index:
    yearly_difference = (corolla_medians.loc[2018] - corolla_medians.loc[2010]) / 8
    print("T4f: Average difference per year (lakh rupees):", round(yearly_difference, 2))
else:
    print("T4f: Cannot calculate the average because 2010 or 2018 is missing.")

# T4g: Five most expensive advertisements
print("T4g: Five most expensive ads:")
top_expensive = df.sort_values("Price", ascending=False)[["Brand", "Model", "Year", "Price", "Registered City"]].head(5)
display(top_expensive)
print("T4g: The table above shows the five highest advertised prices.")

# T4h: Number of ads and median price for Cash versus Installment/Leasing
print("T4h: Number of ads by transaction type:")
print(df["Transaction Type"].value_counts())
print("T4h: Median price in lakh by transaction type:")
print(df.groupby("Transaction Type")["Price_lakh"].median().sort_values(ascending=False))
transaction_counts = df["Transaction Type"].value_counts()
transaction_medians = df.groupby("Transaction Type")["Price_lakh"].median()
cash_count = int(transaction_counts.get("Cash", 0))
installment_count = int(transaction_counts.get("Installment/Leasing", 0))
cash_median = transaction_medians.get("Cash", float("nan"))
installment_median = transaction_medians.get("Installment/Leasing", float("nan"))
print("T4h: Cash advertisement count:", cash_count)
print("T4h: Installment/Leasing advertisement count:", installment_count)
print("T4h: Cash median price (lakh):", round(cash_median, 2))
print("T4h: Installment/Leasing median price (lakh):", round(installment_median, 2))
```

### T4(f) Written answer

The median advertised price rises from **12.25 lakh** for a 2010 Corolla GLI to **23.20 lakh** for a 2018 Corolla GLI. The difference is **10.95 lakh** over 8 years, which is about **1.37 lakh per year**. In other words, when we compare these year groups, a Corolla GLI that is one year older costs about **1.37 lakh rupees less** on average according to the median prices in this dataset. This is an estimate from advertisements, not a guaranteed depreciation rate for every car.
